```python
import pulp

# 1) Sets: I = {1..14}, J = {1..14}
I = range(1, 15)  # positions
J = range(1, 15)  # possible integer values

# 2) Define the coefficients c[i]
c = {
     1: 18740,
     2: 17605,
     3: 17444,
     4: 16785,
     5: 15819,
     6: 10204,
     7:  3521,
     8:  8081,
     9:  8710,
    10:  4543,
    11:  4013,
    12:   904,
    13:   787,
    14:  2754
}

# 3) Create a PuLP problem (maximize)
problem = pulp.LpProblem("AllDifferentAssignment", pulp.LpMaximize)

# 4) Define binary decision variables y[i,j]
#    y[i,j] = 1 if position i is assigned integer j, else 0
y = pulp.LpVariable.dicts("y", (I, J), cat=pulp.LpBinary)

# 5a) Constraint: Each position i gets exactly one integer j
for i in I:
    problem += pulp.lpSum(y[i][j] for j in J) == 1, f"AssignOneValueToPos_{i}"

# 5b) Constraint: Each integer j is used exactly once among all positions i
for j in J:
    problem += pulp.lpSum(y[i][j] for i in I) == 1, f"UseEachValueOnce_{j}"

# 6) Objective: Maximize sum_{i=1..14} sum_{j=1..14} (c[i] * j * y[i,j])
objective_expr = []
for i in I:
    for j in J:
        objective_expr.append(c[i] * j * y[i][j])

problem += pulp.lpSum(objective_expr), "MaximizeTotalCost"

# 7) Solve the problem
problem.solve(pulp.PULP_CBC_CMD(msg=0))  # CBC is the default solver bundled with PuLP

# 8) Print the results
print(f"Solver status: {pulp.LpStatus[problem.status]}")
print(f"Objective value: {pulp.value(problem.objective)}\n")

print("Assignments (x_i = j):")
for i in I:
    for j in J:
        if pulp.value(y[i][j]) == 1:
            print(f"  x_{i} = {j}")
```
